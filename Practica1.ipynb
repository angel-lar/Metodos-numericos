```python
def producto_cartesiano(A, B):
    resultado = []
    for a in A:
        for b in B:
            resultado.append(a + b)
    return resultado

def potencia_lenguaje(lenguaje, n):
    if n == 0:
        return {"ε"}

    resultado = list(lenguaje)#es él mismo

    # concatenamos n-1 veces los elementos
    for _ in range(n - 1):
        resultado = producto_cartesiano(resultado, lenguaje)

    # acá uso set() para eliminar duplicados si los hay
    return set(resultado)


# 1. POTENCIA DEL LENGUAJE L^n

L1 = ["ba", "bab"]

print("1. Potencias de L = {'ba', 'bab'}")
for n in range(4):
    cadenas = potencia_lenguaje(L1, n)
    print(f"L^{n} ({len(cadenas)} cadenas): {cadenas}")




# 2. CADENAS SOBRE {a, b} DE LONGITUD n=6
print("\n2. Cadenas sobre {a, b} de longitud n = 6")
alfabeto = ["a", "b"]
n = 6

cadenas6= sorted(list(potencia_lenguaje(alfabeto, n)))

print(f"Total de cadenas generadas: {len(cadenas6)}")
print("Cadenas:", cadenas6)


# 3. SIGNIFICADO
print("\n3. Lenguaje de acciones")

def mover_arriba():
    return "Acción: Mover hacia ARRIBA"

def mover_abajo():
    return "Acción: Mover hacia ABAJO"

def mover_izquierda():
    return "Acción: Mover a la IZQUIERDA"

def mover_derecha():
    return "Acción: Mover a la DERECHA"

# Diccionario
lenguaje_comandos = {
    "ba": mover_arriba,
    "bab": mover_abajo,
    "baba": mover_izquierda,
    "babab": mover_derecha
}

print("Ejecutando el significado de cada palabra:")
for palabra, accion in lenguaje_comandos.items():
    print(f"Palabra: '{palabra}' -> {accion()}")
```
